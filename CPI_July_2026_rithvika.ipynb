## 1. Import Pandas

```python
import pandas as pd
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## 2. Load the Excel file

`pd.read_excel()` reads an Excel file and stores it as a Pandas dataframe.

```python
df = pd.read_excel("cpi_july26.xlsx", header=None)
```

```python
df.head(10)
```

## 3. Inspect the raw dataframe

```python
df.shape
```

```python
df.tail(5)
```

```python
df.info()
```

```python
cpi = df.iloc[4:40].copy()
cpi.columns = [
    "s_no", "state", "cpi_rural", "cpi_urban", "cpi_combined",
    "inflation_rural", "inflation_urban", "inflation_combined"
]
cpi.head()
```

## Clean the state names

```python
cpi["state"] = cpi["state"].astype(str).str.strip()
cpi["state"].tolist()
```

## Convert numeric columns

```python
numeric_columns = [
    "s_no", "cpi_rural", "cpi_urban", "cpi_combined",
    "inflation_rural", "inflation_urban", "inflation_combined"
]
for column in numeric_columns:
    cpi[column] = pd.to_numeric(cpi[column], errors="coerce")
cpi.dtypes
```

## Check for missing values

```python
cpi.isnull().sum()
```

The source notes that Chandigarh has no rural market. Its rural CPI and rural inflation values are therefore shown as zero in the source and should not be interpreted as a normal measured rural observation.

## Check for duplicate rows

```python
cpi.duplicated().sum()
```

## Check the cleaned dataset

```python
cpi.shape
```

```python
cpi.head()
```

```python
cpi.tail()
```

```python
cpi.info()
```

## Save the cleaned data as a CSV

```python
cpi.to_csv("cpi_july_2026_cleaned.csv", index=False)
```

## Read the cleaned CSV back into Pandas

```python
cpi_clean = pd.read_csv("cpi_july_2026_cleaned.csv")
cpi_clean.head()
```

```python
cpi_clean.shape
```

## Basic statistics

```python
cpi_clean.describe()
```

```python
cpi_clean[["state", "cpi_combined", "inflation_combined"]]
```

## Sort states by combined inflation

```python
cpi_clean.sort_values(by="inflation_combined", ascending=False).head(10)
```

## Find the states with the lowest inflation

```python
cpi_clean.sort_values(by="inflation_combined", ascending=True).head(10)
```

## Filter states with inflation above 5%

```python
high_inflation = cpi_clean[cpi_clean["inflation_combined"] > 5]
high_inflation[["state", "inflation_combined"]].sort_values(by="inflation_combined", ascending=False)
```

##  Filter states with inflation below 3%

```python
low_inflation = cpi_clean[cpi_clean["inflation_combined"] < 3]
low_inflation[["state", "inflation_combined"]].sort_values(by="inflation_combined")
```

##  Calculate the rural-urban inflation gap

A positive value means rural inflation is higher than urban inflation. A negative value means urban inflation is higher.

```python
cpi_clean["rural_urban_inflation_gap"] = cpi_clean["inflation_rural"] - cpi_clean["inflation_urban"]
cpi_clean[["state", "inflation_rural", "inflation_urban", "rural_urban_inflation_gap"]].head()
```

## Find the largest rural-urban inflation gaps

```python
cpi_clean[["state", "inflation_rural", "inflation_urban", "rural_urban_inflation_gap"]].sort_values(by="rural_urban_inflation_gap", ascending=False).head(10)
```

## Find where urban inflation is higher

```python
cpi_clean[["state", "inflation_rural", "inflation_urban", "rural_urban_inflation_gap"]].sort_values(by="rural_urban_inflation_gap", ascending=True).head(10)
```

##  Compare rural and urban averages

```python
print("Average rural inflation:", round(cpi_clean["inflation_rural"].mean(), 2))
print("Average urban inflation:", round(cpi_clean["inflation_urban"].mean(), 2))
print("Average combined inflation:", round(cpi_clean["inflation_combined"].mean(), 2))
```

##  Compare Tamil Nadu with the rest of the dataset

```python
tamil_nadu = cpi_clean[cpi_clean["state"].str.contains("Tamil Nadu", case=False, na=False)]
tamil_nadu
```

```python
print("Tamil Nadu combined CPI:", tamil_nadu["cpi_combined"].iloc[0])
print("Tamil Nadu rural inflation:", tamil_nadu["inflation_rural"].iloc[0])
print("Tamil Nadu urban inflation:", tamil_nadu["inflation_urban"].iloc[0])
print("Tamil Nadu combined inflation:", tamil_nadu["inflation_combined"].iloc[0])
```

##  Rank states by combined inflation

```python
cpi_clean["inflation_rank"] = cpi_clean["inflation_combined"].rank(method="min", ascending=False)
cpi_clean[["state", "inflation_combined", "inflation_rank"]].sort_values("inflation_rank").head(10)
```

##  Identify states above the average inflation rate

```python
average_inflation = cpi_clean["inflation_combined"].mean()
above_average = cpi_clean[cpi_clean["inflation_combined"] > average_inflation]
print("Average combined inflation:", round(average_inflation, 2))
above_average[["state", "inflation_combined"]].sort_values("inflation_combined", ascending=False)
```

##  Final data-quality check

```python
print("Rows:", cpi_clean.shape[0])
print("Columns:", cpi_clean.shape[1])
print("Duplicate rows:", cpi_clean.duplicated().sum())
print("Missing values:")
print(cpi_clean.isnull().sum())
```

## 26. Save the final cleaned dataset

```python
cpi_clean.to_csv("cpi_july_2026_cleaned_final.csv", index=False)
```

# Story Pitch


## Tamil Nadu Rural Inflation Outpaces Urban Rate in July

Tamil Nadu's combined retail inflation stood at **5.44% in July 2026**, with rural inflation at **5.59%**, compared with **5.31% in urban areas**. The 0.28-percentage-point gap puts Tamil Nadu among the states with the highest inflation rates in the country and raises questions about what is driving price pressures outside the state's urban centres.

### Key Data Points

* **Tamil Nadu rural inflation:** 5.59%
* **Tamil Nadu urban inflation:** 5.31%
* **Rural-urban gap:** 0.28 percentage points
* **Combined inflation:** 5.44%
* **Rural CPI index:** 108.81
* **Urban CPI index:** 108.74
* **Combined CPI index:** 108.77
* **National combined inflation:** 3.98%
* **Tamil Nadu's position:** 3rd highest combined inflation among 36 states and Union Territories

### Story Angle

The July CPI data shows that **rural Tamil Nadu faced higher inflation than urban Tamil Nadu**, even though the rural and urban CPI indices were almost identical.

The data itself does not explain the gap. That creates the reporting question: **what is causing rural price pressures to run ahead of urban inflation?**

Possible explanations to investigate include differences in food prices, fuel costs, distribution and transportation costs, and changes in the prices of goods that make up rural household consumption baskets. These should be tested against category-level data and expert interviews rather than assumed from the headline numbers.

### A sustained rural-urban inflation gap could have implications for household purchasing power, particularly for agricultural workers and other rural households whose incomes may not rise as quickly as consumer prices. It could also provide a more granular picture of inflation in Tamil Nadu than the headline state-level number: the average inflation rate may mask different price experiences across rural and urban households.

* Economists specialising in inflation and rural consumption
* Agricultural economists for food-price trends
* Retailers and distributors in rural Tamil Nadu
* Rural households and agricultural workers
