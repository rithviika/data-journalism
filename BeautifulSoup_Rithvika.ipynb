## Scraping the quotes


counting the number of quotes first

```python
quotes = soup.find_all("div", class_="quote")

print(len(quotes))
```

```python
#10 quotes
```

extracting the 10 quotes (using for loop here)

```python
for quote in quotes:
    text = quote.find("span", class_="text").get_text(strip=True)
    print(text)
```

Now scraping the quote with the author

```python
for quote in quotes:
    text = quote.find("span", class_="text").get_text(strip=True)
    author = quote.find("small", class_="author").get_text(strip=True)
    
    print(text, "-", author)
    print("_" * 120) #to draw a line between the quotes
```

adding tags

```python
for quote in quotes:
    text = quote.find("span", class_="text").get_text(strip=True)
    author = quote.find("small", class_="author").get_text(strip=True)
    
    tags = quote.find_all("a", class_="tag")
    tag_list = [tag.get_text(strip=True) for tag in tags]
    
    print(text, "-", author)
    print("Tags:", tag_list)
    print("_" * 120) #to draw a line between the quotes
```

Creating a dataframe

```python
data = [] 

for quote in quotes:
    text = quote.find("span", class_="text").get_text(strip=True)
    author = quote.find("small", class_="author").get_text(strip=True)
    
    tags = quote.find_all("a", class_="tag")
    tag_list = [tag.get_text(strip=True) for tag in tags]
    
    data.append({
        "Quote": text,
        "Author": author,
        "Tags": ", ".join(tag_list)
    })
```

```python
df = pd.DataFrame(data)
df
```
